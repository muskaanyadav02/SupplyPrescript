```python
import pandas as pd
import numpy as np
import sqlite3
import os

pd.set_option("display.max_columns", None)

print("Libraries imported successfully!")
```

```python
input_file = "../data/raw/DataCoSupplyChainDataset.csv"

df = pd.read_csv(
    input_file,
    encoding="latin1"
)

print("Dataset loaded successfully!")
print("Shape:", df.shape)
```

```python
df.head()
```

```python
df.info()
```

```python
missing = df.isnull().sum()

missing[missing > 0]
```

```python
duplicates = df.duplicated().sum()

print("Duplicate Rows:", duplicates)
```

```python
df_clean = df.copy()

print("Copy created successfully!")
print(df_clean.shape)
```

```python
columns_to_drop = [
    "Customer Email",
    "Customer Password",
    "Product Image"
]

df_clean.drop(
    columns=columns_to_drop,
    inplace=True
)

print("Columns removed successfully!")
print(df_clean.shape)
```

```python
df_clean.columns = (
    df_clean.columns
    .str.strip()
    .str.lower()
    .str.replace(" ", "_")
    .str.replace("(", "", regex=False)
    .str.replace(")", "", regex=False)
)

print("Column names standardized.")
```

```python
df_clean.columns.tolist()
```

```python
print(
    df_clean[
        [
            "order_date_dateorders",
            "shipping_date_dateorders"
        ]
    ].dtypes
)
```

```python
date_columns = [
    "order_date_dateorders",
    "shipping_date_dateorders"
]

for col in date_columns:
    df_clean[col] = pd.to_datetime(
        df_clean[col],
        errors="coerce"
    )

print("Date conversion completed!")

print(df_clean[date_columns].dtypes)
```

```python
print(
    "Invalid Order Dates:",
    df_clean["order_date_dateorders"].isna().sum()
)

print(
    "Invalid Shipping Dates:",
    df_clean["shipping_date_dateorders"].isna().sum()
)
```

```python
before_memory = (
    df_clean.memory_usage(deep=True)
    .sum()
    / (1024 ** 2)
)

print(f"Memory Before Optimization : {before_memory:.2f} MB")
```

```python
int_columns = df_clean.select_dtypes(
    include=["int64"]
).columns

for col in int_columns:
    df_clean[col] = pd.to_numeric(
        df_clean[col],
        downcast="integer"
    )

float_columns = df_clean.select_dtypes(
    include=["float64"]
).columns

for col in float_columns:
    df_clean[col] = pd.to_numeric(
        df_clean[col],
        downcast="float"
    )
```

```python
for col in df_clean.select_dtypes(include="object").columns:

    if df_clean[col].nunique() < 100:

        df_clean[col] = df_clean[col].astype("category")
```

```python
after_memory = (
    df_clean.memory_usage(deep=True)
    .sum()
    / (1024 ** 2)
)

print(f"Memory Before Optimization : {before_memory:.2f} MB")
print(f"Memory After Optimization  : {after_memory:.2f} MB")
print(f"Memory Saved               : {before_memory-after_memory:.2f} MB")
```

```python
os.makedirs("../data/processed", exist_ok=True)

df_clean.to_csv(
    "../data/processed/clean_supply_chain.csv",
    index=False
)

print("Clean dataset saved successfully!")
```

```python
conn = sqlite3.connect("../database/supply_chain.db")

df_clean.to_sql(
    "supply_chain",
    conn,
    if_exists="replace",
    index=False
)

print("Rows Loaded:", len(df_clean))

conn.close()

print("SQLite database created successfully!")
```
